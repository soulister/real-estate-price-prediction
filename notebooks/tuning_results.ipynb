```python
X_train = prep_with_features_eng(train_df)
cat_categories = fit_cat_categories(X_train)
X_test = prep_with_features_eng(test_df, cat_categories=cat_categories)
y_train, y_test = X_train['price_doc'], X_test['price_doc']
X_train, X_test = X_train[selected_features], X_test[selected_features]
```

```python
model_xgb = XGBRegressor(**best_params_xgb,
                    eval_metric =  "rmse",
                    tree_method = "hist",
                    objective = 'reg:squarederror',
                    enable_categorical = True,
                    random_state = 42,
                    n_jobs = -1,)
model_xgb.fit(X_train,
              y_train,
              verbose=200)
```

```python
xgb_pred_log = model_xgb.predict(X_test)
xgb_pred = np.expm1(xgb_pred_log)

score_xgb = rmsle(np.expm1(y_test), xgb_pred)
score_xgb
```

```python
cat_features = X_train.select_dtypes(
            include="category"
            ).columns.to_list()

model_cat = CatBoostRegressor(**best_params_cat,
                        eval_metric = "RMSE",
                        loss_function = 'RMSE',
                        random_seed = 42,
                        verbose = 200)

model_cat.fit(X_train,
            y_train,
            cat_features=cat_features,
            use_best_model=True)
```

```python
model_lgb = LGBMRegressor(**best_params_lgb,
                    objective = 'regression',
                    min_child_samples = 20,
                    random_state = 42,
                    n_jobs = -1,
                    verbose = 200)

model_lgb.fit(X_train,
              y_train,
              eval_metric="rmse",
              categorical_feature=cat_features)
```

```python
lgb_pred_log = model_lgb.predict(X_test)
lgb_pred = np.expm1(lgb_pred_log)

score_lgb = rmsle(np.expm1(y_test), lgb_pred)
score_lgb
```

```python
oof_pred_xgb = oof_pred_xgb.reset_index(drop=True)
oof_pred_lightgbm = oof_pred_lightgbm.reset_index(drop=True)
oof_pred_cat = oof_pred_cat.reset_index(drop=True)


pred_journal = oof_pred_xgb.merge(
        oof_pred_lightgbm[["id", "fold", "val_block", "lightgbm_100_tuned_pred", "lightgbm_100_tuned_error"]],
        on=["id", "fold", "val_block"],
        how="left"
        ).merge(
        oof_pred_cat[["id", "fold", "val_block", "catboost_100_tuned_pred", "catboost_100_tuned_error"]],
        on=["id", "fold", "val_block"],
        how="left"
        )
pred_journal
```

```python
stack_df = pred_journal[['catboost_100_tuned_pred', 'xgboost_100_tuned_pred', 'lightgbm_100_tuned_pred', 'y_true']]
```

```python
from sklearn.linear_model import RidgeCV

pred_cols = ['catboost_100_tuned_pred', 'xgboost_100_tuned_pred', 'lightgbm_100_tuned_pred']

X_meta = np.log1p(stack_df[pred_cols])
y_meta = np.log1p(stack_df["y_true"])
```

```python
def alpha_objective(trial):
    meta_model = RidgeCV(
        alphas=trial.suggest_float("alpha", 1e-8, 1e5, log=True)
    )

    meta_model.fit(X_meta, y_meta)
    holdout_stack = pd.DataFrame({'y_true': y_test, 'catboost_100_tuned_pred': cat_pred_log, 'xgboost_100_tuned_pred': xgb_pred_log, 'lightgbm_100_tuned_pred': lgb_pred_log})
    X_meta_test = holdout_stack[pred_cols]
    y_meta_test = holdout_stack["y_true"]

    stack_pred_log = meta_model.predict(X_meta_test)
    stack_pred = np.expm1(stack_pred_log)

    ridge_stack_score = rmsle(
        np.expm1(y_meta_test),
        stack_pred
    )

    return ridge_stack_score
```

```python
study_ridge = optuna.create_study(direction='minimize',
        sampler=optuna.samplers.TPESampler(seed=456))
study_ridge.optimize(alpha_objective, n_trials=500, show_progress_bar=True)
```
